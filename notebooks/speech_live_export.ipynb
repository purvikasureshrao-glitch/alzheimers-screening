# Speech - final live-app network (Whisper features) [OURS]
**Runs on:** Colab (CPU is enough). **Uploads (3 small files, no audio, no transcripts):** `speech_code.tar` (re-made), `speech_whisper_features.csv`, `speech_whisper_oof.csv` (for the labels). Fits the 5-feature Disfluency network on all 108 subjects and exports ONNX. **Not an evaluation**; the only accuracy statement is the Whisper `small` 5-fold CV (FINDINGS 9b: 0.639, 0.546-0.731).

```python
import os, sys, subprocess
os.makedirs('/content/Alzheimers', exist_ok=True); os.makedirs('/content/out', exist_ok=True)
subprocess.run(['tar','-xf','/content/speech_code.tar','-C','/content/Alzheimers'], check=True)
subprocess.run([sys.executable,'-m','pip','install','-q','skl2onnx','onnxruntime','onnx','scikit-learn'], check=True)
sys.path.insert(0,'/content/Alzheimers')
import numpy as np, pandas as pd
from speech.export import export_whisper
from speech.whisper_features import WHISPER_FEATURE_NAMES
f = pd.read_csv('/content/speech_whisper_features.csv'); o = pd.read_csv('/content/speech_whisper_oof.csv')
assert (f.id.values == o.id.values).all(), 'id order differs'
X = f[WHISPER_FEATURE_NAMES].values; y = o.label.values
print(X.shape, np.bincount(y), 'all finite:', bool(np.isfinite(X).all()))
export_whisper(X, y, '/content/out', seed=0, device='cpu')
```
